```python
for col in songs.columns:
    if songs[col].dtype == 'str':
        songs[col] = songs[col].str.strip()
```

Creating new column for popularity and based on streme table changing the popularity scores

```python
#song_streams
song_streams = (
    streams.groupby("Song_ID")
    .size()
    .reset_index(name="Stream_Count")
)
```

```python
songs = songs.merge(
    song_streams,
    on="Song_ID",
    how="left"
)
```

```python
songs.info()
```

```python
songs["Stream_Count"] = (
    songs["Stream_Count"]
    .fillna(0)
    .astype(int)
)
```

```python
songs.info()
```

```python
songs["Stream_Count"].describe()
```

```python
songs["Stream_Count"].value_counts().sort_index()
```

```python
songs["Popularity"] = (
    songs["Stream_Count"]
    .rank(method="average", pct=True) * 100
).round().clip(1, 100).astype(int)
```

```python
songs[
    ["Song_ID", "Stream_Count", "Popularity"]
].sort_values(
    "Popularity",
    ascending=False
).head(20)
```

Save all cleaned tables into a new Excel file

```python
output_file="C:\\BEAT-FINAL-PROJECT\\Data\\Cleaned\\Spotify_Dataset_Cleaned.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    users.to_excel(writer, sheet_name="Users", index=False)
    artists.to_excel(writer, sheet_name="Artists", index=False)
    songs.to_excel(writer, sheet_name="Songs", index=False)
    playlists.to_excel(writer, sheet_name="Playlists", index=False)
    streams.to_excel(writer, sheet_name="Streams", index=False)

print(f"Cleaned dataset saved to: {output_file}")



```
